# Notebook summary

This notebook initializes the Unity Catalog layout for the Cosmos-to-Databricks POC.

* Creates the `cosmos_poc` catalog on the specified ADLS managed location.
* Creates three schemas:
  * `raw` for untouched Cosmos change-feed JSON
  * `bronze` for Auto Loader Delta output plus ingestion metadata
  * `ops` for checkpoints and schema-tracking assets
* Creates two volumes:
  * `cosmos_poc.raw.cosmos_landing` for landed JSON Lines files
  * `cosmos_poc.ops.checkpoints` for change-feed and Auto Loader checkpoint data
* Verifies the landing volume definition so you can confirm storage is backed by ADLS.

```sql
CREATE CATALOG IF NOT EXISTS cosmos_poc
MANAGED LOCATION 'abfss://poc-uc@stdbxpocsm.dfs.core.windows.net/cosmos_poc'
COMMENT 'Cosmos to Databricks POC - raw and bronze layers';
```

```sql
USE CATALOG cosmos_poc;
CREATE SCHEMA IF NOT EXISTS raw    COMMENT 'Landing zone: untouched JSON from Cosmos change feed';
CREATE SCHEMA IF NOT EXISTS bronze COMMENT 'Auto Loader output: raw data as Delta + ingestion metadata';
CREATE SCHEMA IF NOT EXISTS ops    COMMENT 'Operational: checkpoints and schema tracking';
```

```sql
CREATE VOLUME IF NOT EXISTS cosmos_poc.raw.cosmos_landing
  COMMENT 'Raw JSON Lines files written by the extract task';
CREATE VOLUME IF NOT EXISTS cosmos_poc.ops.checkpoints
  COMMENT 'Change feed and Auto Loader checkpoints, Auto Loader schema location';
```

```sql
DESCRIBE VOLUME cosmos_poc.raw.cosmos_landing;
```
